```python
import pandas as pd
df = pd.read_csv("vitals_raw_data.csv")
```

```python
df
```

```python
df.columns
```

```python
df["STUDYID"] = "STUDY1"
```

```python
df["USUBJID"] = df["STUDYID"] + "-" + df["PATNUM"].astype(str)
```

```python
df
```

```python
# Define columns to melt
value_vars = ["SYS_BP", "DIA_BP", "PULSE", "TEMP"]
```

```python
vs = df.melt(
    id_vars=["STUDYID", "USUBJID", "SUBPOS", "TEMPLOC", "LOC"],
    value_vars=value_vars,
    var_name="RAW_TEST",
    value_name="VSORRES"
)
```

```python
vs
```

```python
vs = vs.dropna(subset=["VSORRES"])
```

```python
# Map RAW_TEST → VSTESTCD
test_map = {
    "SYS_BP": "SYSBP",
    "DIA_BP": "DIABP",
    "PULSE": "PULSE",
    "TEMP": "TEMP"
}
vs["VSTESTCD"] = vs["RAW_TEST"].map(test_map)
```

```python
vs
```

```python
unit_map = {
    "SYSBP": "mmHg",
    "DIABP": "mmHg",
    "PULSE": "beats/min",
    "TEMP": "C"
}
vs["VSORRESU"] = vs["VSTESTCD"].map(unit_map)
```

```python
vs
```

```python
# Standard result
vs["VSSTRESC"] = vs["VSORRES"]
vs["VSSTRESU"] = vs["VSORRESU"]

# Position
vs["VSPOS"] = vs["SUBPOS"]
```

```python
vs
```

```python
vs["VSLOC"] = vs.apply(
    lambda x: x["TEMPLOC"] if x["VSTESTCD"] == "TEMP" else x["LOC"],
    axis=1
)
```

```python
vs
```

```python
vs = vs[
    [
        "STUDYID",
        "USUBJID",
        "VSTESTCD",
        "VSORRES",
        "VSORRESU",
        "VSSTRESC",
        "VSSTRESU",
        "VSPOS",
        "VSLOC",
    ]
]
```

```python
vs
```

```python
vs.to_csv("vs.csv", index=False)
```
